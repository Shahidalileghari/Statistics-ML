## Learning objectives
- Use conditional probability
- Apply Bayes' theorem
- Interpret base rates

Bayes' theorem reverses a conditional probability by combining the prior prevalence with test performance. Natural frequencies often make the result easier to understand.

## Worked example


```python
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90
p_positive = sensitivity*prevalence + (1-specificity)*(1-prevalence)
p_disease_given_positive = sensitivity*prevalence/p_positive
print('P(positive):', round(p_positive, 4))
print('P(disease | positive):', round(p_disease_given_positive, 4))
people = 10_000
true_positive = people*prevalence*sensitivity
false_positive = people*(1-prevalence)*(1-specificity)
print('True positives:', round(true_positive))
print('False positives:', round(false_positive))
```
