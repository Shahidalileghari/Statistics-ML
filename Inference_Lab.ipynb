# Inference Lab: Confidence Intervals and Tests

## Learning objectives
- Build a t confidence interval
- Run a one-sample t-test
- Interpret results in context

A confidence interval expresses uncertainty from sampling under a method and its assumptions. A p-value measures how incompatible the observed data are with a null model; it is not the probability the null is true.

## Worked example


```python
import numpy as np
from scipy import stats
scores = np.array([72, 75, 78, 81, 69, 88, 76, 74, 83, 79, 71, 85], dtype=float)
ci = stats.t.interval(.95, df=len(scores)-1, loc=scores.mean(), scale=stats.sem(scores))
test = stats.ttest_1samp(scores, popmean=75)
print('mean:', scores.mean())
print('95% CI:', ci)
print('t statistic:', test.statistic)
print('p-value:', test.pvalue)
```
