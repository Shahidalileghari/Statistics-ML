Descriptive statistics summarize the observed sample. The mean is sensitive to extreme values; the median is more robust. Sample variance uses n−1 in the denominator.

## Worked example


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([1, 2, 2, 3, 4, 5, 20], dtype=float)
print('mean:', x.mean())
print('median:', np.median(x))
print('sample SD:', x.std(ddof=1))
print('Q1/Q3:', np.quantile(x, [.25, .75]))
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(x, bins=7, edgecolor='black')
ax.set(xlabel='Value', ylabel='Frequency', title='Small sample with an extreme value')
plt.show()
```
